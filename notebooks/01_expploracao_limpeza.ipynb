```python
import pandas as pd

caminho_arquivo = "../data/raw/itens_raw.csv"

df = pd.read_csv(caminho_arquivo)

df.head()
```

```python
df.info()
```

```python
df['categorias_json']
```

```python
# Colunas selecionadas para visão simplificada dos itens.

colunas_selecionadas = [
    'item_id',
    'nome',
    'descricao_texto',
    'categorias_json',
    'preco_total_ouro'
]

df_itens=df[colunas_selecionadas].copy()

df_itens.head()
```

```python
df_itens.info()
```

```python
import json

# Converte as strings formatadas em JSON para listas reais e expande os elementos em novas linhas

df_itens['categorias_json'] = df_itens['categorias_json'].apply(json.loads)
df_itens_explodidos = df_itens[['item_id','categorias_json']].explode('categorias_json')
df_itens_explodidos
```

```python
df_itens_explodidos['categorias_json'] = df_itens_explodidos['categorias_json'].fillna("Item Quebrado")
```

```python
categorias_unicas = df_itens_explodidos['categorias_json'].unique()
categorias_unicas
```

```python
df_categorias_unicas = pd.DataFrame(categorias_unicas, columns=['nome_categoria']).reset_index()
df_categorias_unicas = df_categorias_unicas.rename(columns={'index':'id_categoria'})
df_categorias_unicas['id_categoria'] += 1
df_categorias_unicas.head()
```

```python
df_item_categoria = pd.merge(
    df_itens_explodidos,
    df_categorias_unicas,
    left_on='categorias_json',
    right_on='nome_categoria',
    how='inner'
)

df_item_categoria=df_item_categoria[['item_id','id_categoria']]

df_item_categoria = df_item_categoria.drop_duplicates().reset_index(drop=True)
df_item_categoria
```

```python
colunas_finais_item = ['item_id','nome','preco_total_ouro','descricao_texto']
df_tabela_item = df_itens[colunas_finais_item].copy()

df_tabela_item = df_tabela_item.rename(columns={
    'preco_total_ouro':'preco_unitario',
    'descricao_texto':'descricao'
})

df_tabela_item = df_tabela_item.drop_duplicates(subset=['item_id']).reset_index(drop=True)

df_tabela_item.head()
```

```python
elos = [
    'Ferro', 'Bronze', 'Prata', 'Ouro', 
    'Platina', 'Esmeralda', 'Diamante', 
    'Mestre', 'Grão-Mestre', 'Desafiante'
]

df_elo = pd.DataFrame(elos,columns=['elo']).reset_index()
df_elo = df_elo.rename(columns = {'index': 'id_elo'})
df_elo['id_elo'] += 1

df_elo
```

```python
import random

random.seed(42)

nicks = [
    'Faker', 'ShowMaker', 'Chovy', 'Caps', 'Rekkles',
    'BrTT', 'Kami', 'Tinowns', 'Robo', 'Cariok',
    'Titan', 'Aegis', 'Ceos', 'Route', 'Dynquedo',
    'Jojo', 'TitaN', 'Gvoy', 'Damage', 'Wizer',
    'Crocodilo', 'Smeb', 'BeryL', 'Deft', 'Keria'
]
```

```python
dados_jogadores = []
for index, nick in enumerate (nicks,start=1):
    dados_jogadores.append({
        'id_jogador' : index,
        'nick' : nick,
        'id_elo': random.randint(1,10)
    })
df_jogador = pd.DataFrame(dados_jogadores)
df_jogador.head(10)    
```

```python
partidas = []
resultados = ['Vitoria','Derrota']
for i in range(1,21):
    partidas.append({
        'id_partida': i,
        'duracao_minutos': random.randint(18,45),
        'resultado': random.choice(resultados)
    })

df_partida = pd.DataFrame(partidas)
df_partida    
```

```python
itens_ids_validos = df_tabela_item['item_id'].tolist()
jogadores_ids_validos = df_jogador['id_jogador'].tolist()
partidas_ids_validos = df_partida['id_partida'].tolist()

compras = []
id_compra = 1

for id_partida in partidas_ids_validos:
    participantes = random.sample(jogadores_ids_validos, 10)

    for jogador in participantes:
        quantidade_itens = random.randint(2,5)
        for _ in range(quantidade_itens):
            compras.append({
                'id_compra':id_compra,
                'id_jogador':jogador,
                'id_item':random.choice(itens_ids_validos),
                'id_partida':id_partida,
                'minuto_compra':random.randint(1,35)
            })
            id_compra += 1

df_compra = pd.DataFrame(compras)
df_compra.head()
```

```python
import os

caminho_processados = '../data/processed'
os.makedirs(caminho_processados, exist_ok=True)

df_elo.to_csv(f"{caminho_processados}/elo.csv", index=False)
df_jogador.to_csv(f"{caminho_processados}/jogador.csv", index=False)
df_tabela_item.to_csv(f"{caminho_processados}/item.csv", index=False)
df_categorias_unicas.to_csv(f"{caminho_processados}/categoria.csv", index=False)
df_item_categoria.to_csv(f"{caminho_processados}/item_categoria.csv", index=False)
df_partida.to_csv(f"{caminho_processados}/partida.csv", index=False)
df_compra.to_csv(f"{caminho_processados}/compra.csv", index=False)

print("Todos os ficheiros CSV foram esportados para data/processed/!")
```
