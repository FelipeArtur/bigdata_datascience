## 1. Introdução

Este trabalho tem como objetivo demonstrar, na prática, o processo de migração de dados de um
banco de dados relacional (PostgreSQL) para um banco de dados NoSQL orientado a documentos
(MongoDB). Utilizamos como cenário uma loja virtual do jogo League of Legends, na qual jogadores
adquirem itens (skins, campeões, passes) ao longo de suas partidas.

Com o crescimento do volume e da diversidade dos dados modernos, os bancos relacionais passam a
enfrentar limitações de flexibilidade e desempenho em determinados cenários. Os bancos NoSQL,
como o MongoDB, surgem como alternativa, armazenando os dados em documentos flexíveis
(semelhantes a JSON) que permitem representar informações complexas de forma mais natural.

Ao longo do relatório, apresentamos o cenário relacional de origem, o modelo de documentos
idealizado, as decisões de modelagem (embedding versus referencing) com suas justificativas
técnicas, e uma avaliação crítica das vantagens e limitações da migração.

## 2. Descrição do cenário

O cenário escolhido é a **loja virtual do jogo League of Legends (LoL)**, um jogo online no qual
milhões de jogadores adquirem itens virtuais — skins, campeões e passes de batalha — ao longo de
suas partidas. Esse contexto gera um grande volume de compras associadas a jogadores, itens e
partidas, o que o torna adequado para demonstrar a modelagem relacional e a migração para NoSQL.

### Modelo relacional de origem

O banco relacional foi implementado no PostgreSQL, com **6 tabelas** e mais de 70 registros:

| Tabela | Descrição | Campos principais |
|---|---|---|
| `elo` | Elos/ranques do jogo | id_elo, elo |
| `jogador` | Jogadores da loja | id_jogador, nick, regiao, id_elo |
| `categoria_item` | Categorias dos itens | id_categoria_item, categoria_item |
| `item` | Itens à venda | id_item, item, preco_unitario, id_categoria_item |
| `partida` | Partidas jogadas | id_partida, data_partida, duracao_min, resultado |
| `compra` | Compras realizadas | id_compra, id_jogador, id_item, id_partida, data_compra, quantidade |

### Relacionamentos

Os vínculos entre as tabelas são estabelecidos por chaves estrangeiras (FK):

- **elo (1) → (N) jogador**: cada jogador pertence a um elo; um elo tem vários jogadores.
- **categoria_item (1) → (N) item**: cada item pertence a uma categoria.
- **jogador (1) → (N) compra**: cada jogador realiza várias compras.
- **item (1) → (N) compra**: cada compra refere-se a um item.
- **partida (1) → (N) compra**: cada compra ocorre no contexto de uma partida.

A tabela `compra` é a tabela central (fato), conectando jogadores, itens e partidas. No modelo
relacional, obter a visão completa de uma compra (nick do jogador, nome do item, elo etc.) exige
combinar várias tabelas com operações de JOIN — o que motiva, na etapa seguinte, a consolidação
desses dados em documentos.

````python
%%sql
-- Apaga tabelas antigas, se existirem (pra poder rodar de novo sem erro)
DROP TABLE IF EXISTS compra CASCADE;
DROP TABLE IF EXISTS item CASCADE;
DROP TABLE IF EXISTS jogador CASCADE;
DROP TABLE IF EXISTS partida CASCADE;
DROP TABLE IF EXISTS categoria_item CASCADE;
DROP TABLE IF EXISTS elo CASCADE;

-- 1. Elos do jogo (Ferro, Bronze, Prata...)
CREATE TABLE elo (
    id_elo SERIAL PRIMARY KEY,
    elo VARCHAR(25) NOT NULL UNIQUE
);

-- 2. Jogadores (adicionamos a REGIÃO)
CREATE TABLE jogador (
    id_jogador SERIAL PRIMARY KEY,
    nick VARCHAR(50) NOT NULL UNIQUE,
    regiao VARCHAR(10) NOT NULL,
    id_elo INT REFERENCES elo(id_elo)
);

-- 3. Categorias de item (Skin, Campeão, Passe...)
CREATE TABLE categoria_item (
    id_categoria_item SERIAL PRIMARY KEY,
    categoria_item VARCHAR(50) NOT NULL UNIQUE
);

-- 4. Itens da loja
CREATE TABLE item (
    id_item SERIAL PRIMARY KEY,
    item VARCHAR(50) NOT NULL UNIQUE,
    preco_unitario INT NOT NULL,
    id_categoria_item INT REFERENCES categoria_item(id_categoria_item)
);

-- 5. Partidas (enriquecemos com data, duração e resultado)
CREATE TABLE partida (
    id_partida SERIAL PRIMARY KEY,
    data_partida DATE NOT NULL,
    duracao_min INT,
    resultado VARCHAR(10)
);

-- 6. Compras (adicionamos DATA e QUANTIDADE)
CREATE TABLE compra (
    id_compra SERIAL PRIMARY KEY,
    id_jogador INT REFERENCES jogador(id_jogador),
    id_item INT REFERENCES item(id_item),
    id_partida INT REFERENCES partida(id_partida),
    data_compra DATE NOT NULL,
    quantidade INT NOT NULL DEFAULT 1
);
````

## 3. Modelo NoSQL idealizado

Nesta etapa, o modelo relacional foi transformado em um modelo de documentos do MongoDB, com o
objetivo de aproveitar a flexibilidade dos documentos JSON, reduzir a necessidade de JOINs e
agrupar as informações que são lidas em conjunto.

### 3.1 Estrutura e organização dos dados

As 6 tabelas relacionais foram reorganizadas em **3 coleções**:

| Coleção | Origem | Estratégia |
|---|---|---|
| `jogadores` | jogador + elo + compra (+ item) | Embedding |
| `itens` | item + categoria_item | Referencing (catálogo) |
| `partidas` | partida | Referencing |

Cada documento de **jogadores** representa um jogador e traz, embutidos, o seu elo e a lista de
compras. Cada compra guarda um "retrato" do item (nome, categoria, preço) no momento da compra,
além de quantidade, data e uma referência à partida. A coleção **itens** funciona como catálogo
de produtos, e **partidas** guarda os dados de cada partida — ambas referenciadas pelas compras.

### 3.2 Decisões de modelagem: embedding vs. referencing

**Embedding (incorporação)** — usado para as compras e o elo dentro do jogador:
- As **compras** foram embutidas no jogador porque pertencem a um único jogador e quase sempre
  são lidas junto com ele (o histórico). Isso elimina os JOINs do modelo relacional. Como cada
  jogador tem um número limitado de compras, não há risco de o documento crescer sem controle
  (limite de 16 MB por documento).
- O **elo** foi embutido como texto simples, por ser pequeno, estável e lido junto com o jogador.

**Referencing (referência)** — usado para itens e partidas:
- Os **itens** formam um catálogo compartilhado por muitas compras. Duplicar o item inteiro em
  cada compra geraria redundância e dificultaria a manutenção (ex.: mudança de preço). Por isso
  mantivemos a coleção `itens` separada, guardando apenas um retrato do nome e preço na compra
  para preservar o valor histórico.
- As **partidas** são entidades próprias ligadas a várias compras; referenciá-las pelo
  `id_partida` evita duplicação.

Seguimos a regra prática: **embedding** para dados pequenos, delimitados e lidos em conjunto;
**referencing** para dados grandes, compartilhados ou que mudam de forma independente.

### 3.3 Exemplos de documentos JSON

Coleção `jogadores` (embedding de elo e compras):
```json
{
  "_id": 1,
  "nick": "Faker",
  "regiao": "KR",
  "elo": "Desafiante",
  "compras": [
    { "item": "Skin Lux Elementalista", "categoria": "Skin",
      "preco_unitario": 1820, "quantidade": 1, "total_rp": 1820,
      "data_compra": "2025-09-01", "id_partida": 1 }
  ]
}
```

Coleção `itens` (referencing):
```json
{ "_id": 1, "item": "Skin Lux Elementalista", "preco_unitario": 1820, "categoria": "Skin" }
```

Coleção `partidas` (referencing):
```json
{ "_id": 1, "data_partida": "2025-09-01", "duracao_min": 32, "resultado": "Vitoria" }
```

### 3.4 Escalabilidade e elasticidade

O modelo de documentos é escalável horizontalmente: conforme o volume de jogadores cresce, os
dados podem ser distribuídos entre vários servidores (sharding), usando, por exemplo, a região ou
o id do jogador como chave de particionamento. O schema flexível também permite adicionar novos
campos a documentos específicos sem alterar toda a base — ao contrário do modelo relacional, em
que uma nova coluna afeta a tabela inteira.

### 3.5 Benefícios da modelagem adotada

- **Desempenho de leitura:** o histórico completo de um jogador é obtido em uma única consulta, sem JOINs.
- **Flexibilidade:** o schema evolui sem migrações rígidas de estrutura.
- **Integridade histórica:** o preço do item é preservado no momento da compra.
- **Manutenção:** catálogo de itens e partidas centralizados, evitando redundância.

````python
import pandas as pd
import json

# 1. Ler os CSVs que salvamos
elo        = pd.read_csv("elo.csv")
jogador    = pd.read_csv("jogador.csv")
categoria  = pd.read_csv("categoria_item.csv")
item       = pd.read_csv("item.csv")
partida    = pd.read_csv("partida.csv")
compra     = pd.read_csv("compra.csv")

# 2. "Tradutores" de id -> nome (pra não guardar números soltos)
elo_por_id       = dict(zip(elo["id_elo"], elo["elo"]))
categoria_por_id = dict(zip(categoria["id_categoria_item"], categoria["categoria_item"]))

# 3. Guardar as infos de cada item (nome, preço, categoria) por id
item_info = {}
for _, row in item.iterrows():
    item_info[int(row["id_item"])] = {
        "item": row["item"],
        "preco_unitario": int(row["preco_unitario"]),
        "categoria": categoria_por_id[row["id_categoria_item"]]
    }

# 4. Coleção ITENS (referencing - catálogo separado)
itens_docs = [
    {"_id": int(r["id_item"]), "item": r["item"],
     "preco_unitario": int(r["preco_unitario"]),
     "categoria": categoria_por_id[r["id_categoria_item"]]}
    for _, r in item.iterrows()
]

# 5. Coleção PARTIDAS (referencing - separada)
partidas_docs = [
    {"_id": int(r["id_partida"]), "data_partida": str(r["data_partida"]),
     "duracao_min": int(r["duracao_min"]), "resultado": r["resultado"]}
    for _, r in partida.iterrows()
]

# 6. Coleção JOGADORES (embedding - com elo e compras dentro)
jogadores_docs = []
for _, jog in jogador.iterrows():
    compras_do_jogador = compra[compra["id_jogador"] == jog["id_jogador"]]
    compras_lista = []
    for _, c in compras_do_jogador.iterrows():
        info = item_info[int(c["id_item"])]
        compras_lista.append({
            "item": info["item"],
            "categoria": info["categoria"],
            "preco_unitario": info["preco_unitario"],
            "quantidade": int(c["quantidade"]),
            "total_rp": info["preco_unitario"] * int(c["quantidade"]),
            "data_compra": str(c["data_compra"]),
            "id_partida": int(c["id_partida"])
        })
    jogadores_docs.append({
        "_id": int(jog["id_jogador"]),
        "nick": jog["nick"],
        "regiao": jog["regiao"],
        "elo": elo_por_id[jog["id_elo"]],
        "compras": compras_lista
    })

# 7. Salvar tudo em arquivos JSON
for nome, docs in [("jogadores", jogadores_docs), ("itens", itens_docs), ("partidas", partidas_docs)]:
    with open(f"{nome}.json", "w", encoding="utf-8") as f:
        json.dump(docs, f, ensure_ascii=False, indent=2)
    print(f"✅ {nome}.json — {len(docs)} documentos")

# 8. Mostrar um exemplo de jogador (com compras embutidas)
print("\n--- Exemplo de documento de jogador ---")
print(json.dumps(jogadores_docs[0], ensure_ascii=False, indent=2))
````

## 4. Conclusão e avaliação crítica

A migração da loja de League of Legends do PostgreSQL para o MongoDB demonstrou, na prática, as
diferenças entre o modelo relacional e o modelo orientado a documentos. No modelo relacional, os
dados ficam normalizados em tabelas separadas, garantindo integridade e evitando redundância, mas
exigindo múltiplos JOINs para reconstruir uma visão completa (como todas as compras de um jogador
com seus itens e elos). No modelo de documentos, essas informações foram consolidadas, permitindo
leituras mais rápidas e naturais.

**Vantagens observadas:**
- Leitura do histórico de um jogador em uma única consulta, sem JOINs.
- Schema flexível, que facilita a evolução da estrutura e a escalabilidade horizontal.
- Representação mais próxima da forma como a aplicação consome os dados.

**Limitações e cuidados:**
- O embedding pode gerar alguma duplicação (o nome do item aparece no catálogo e no retrato
  dentro da compra); é preciso avaliar quando a duplicação é aceitável.
- A ausência de JOINs nativos exige planejar bem quais dados incorporar e quais referenciar.
- Documentos muito grandes poderiam se aproximar do limite de 16 MB por documento, exigindo, em
  cenários de altíssimo volume, uma estratégia de referência para o histórico.

**Aprendizados e melhorias futuras:** o principal aprendizado foi que não existe um modelo único
"melhor" — a escolha entre embedding e referencing depende dos padrões de acesso da aplicação.
Como melhoria futura, poderíamos criar índices no MongoDB (por região, por item) para otimizar
consultas e avaliar separar o histórico de compras em uma coleção própria caso o volume por
jogador cresça muito. Esses dados migrados servirão de base para a próxima etapa: o processamento
em larga escala com Apache Spark.
